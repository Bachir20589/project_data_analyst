```python
import pandas as p
import matplotlib.pyplot as mp
import seaborn as sn
with open("etudiant1.csv") as f:
    a=p.read_csv(f)
    # a.info()
    # b=a.describe()
    # c=a.isna().sum()
    med=a["age"].median()
    medp=a["poids"].median()
    a.fillna({"age":med,"poids":medp},inplace=True)
    print(a.duplicated())
    a.drop_duplicates(inplace=True)
    a.loc[5,'age']=20
    d=a.groupby("filiere")["age"].mean()
    print(a["ville"].value_counts())
    print(a.corr(numeric_only=True))
    sn.histplot(data=a,x="poids",bins=10,edgecolor="red")
    mp.show()
    a["filiere"].value_counts().plot(kind="bar")
    mp.xlabel="filiere"
    mp.ylabel="Nombre etudiants"
    sn.heatmap(a.corr(numeric_only=True),annot=True,cmap="coolwarm")
    mp.show()
# print(c)
print(a)
print(d)
mp.show()
a.to_csv("etudiant1.csv",index=False)


```
